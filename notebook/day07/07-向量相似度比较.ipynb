```python
from langchain_ollama import OllamaEmbeddings

# 创建embeddings模型
ollama_embeddings = OllamaEmbeddings(
    # 传入真实有的模型名称
    model='qwen3-embedding:0.6b',
    # 语义维度长度
    dimensions=1024 # 1024维度
)
```

```python
# 批量向量化文本
texts = [
    '我喜欢晓山瑞希',
    '我好想摆烂',
    '我要和千早爱音结婚'
]

# 将文本转化为向量数组
vectors = ollama_embeddings.embed_documents(texts)

# 转换结果
print(f'转换了:{len(vectors)}条,每一条向量的维度:{len(vectors[0])}')

# 向量化单条文本,给用户的输入内容做相似度匹配
user_input_vector = ollama_embeddings.embed_query('我好想躺在家里吃香的喝辣的')
```

```python
# 相似度匹配
# 导入余弦函数模块
import numpy as np

# 比对函数
def cosine_similarity(vector1,vector2):
    """
    余弦相似度函数
    :param vector1: 向量1,用户输入
    :param vector2: 向量2,需要比对的文本片段
    :return: 计算的向量值
    """
    # 点积计算
    dot = np.dot(vector1, vector2)
    # 求模长
    return dot / (np.linalg.norm(vector1) * np.linalg.norm(vector2))

# 循环对比文本片段
for index,v in enumerate(vectors):
    # 调用函数
    result = cosine_similarity(user_input_vector,v)
    print(f"{texts[index]}:", result)
```
