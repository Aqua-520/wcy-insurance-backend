```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain_core.tools import tool
import os

# 读取env文件
load_dotenv()
# 定义工具函数
@tool
def get_weather(location:str):
    """
    传入指定地点,获取当地天气的函数
    :param location: 地点
    :return: 天气查询结果
    """
    return f'{location}当前的天气是:超级无敌螺旋台风大暴雨'

# ✅从环境变量取出真实模型字符串
chat_model = os.getenv("CHAT_MODEL")

# 初始化模型
# langchain会自动根据模型名称推断出厂商,拼接url,设置名称,系统环境变量读取apikey
model = init_chat_model(model=chat_model)

# 创建智能体
my_agent = create_agent(tools=[get_weather],model=model)

# 调用大模型
result = my_agent.invoke({"messages": [("user", "重庆天气怎么样")]})

# 拆解大模型的返回信息
print(result["messages"][-1].content)
```
